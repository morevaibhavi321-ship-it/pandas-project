```python
#6. create age group and calculate number of customers in each group
df['Age Group'] = ['Child' if x <= 17 else 'Adult' if 17 < x <= 59 else 'Senior' for x in df['Age']]
```

```python
#7. Top 10 most purchased item
df['Item Purchased'].value_counts().sort_values(ascending=False)[:10]
```

```python
#8. total and average purchase amount by category
df2 = df.groupby('Category').agg({'Purchase Amount (USD)':['sum','mean']}).reset_index()
```

```python
#9. average purchase amount of subscribed and non subscribed customers
df.groupby('Subscription Status')['Purchase Amount (USD)'].mean()
```

```python
#10. Payment menthod for each customer gender
pd.crosstab(df['Gender'],df['Payment Method'])
```

```python
#11. customer groups by location and category based on total purchase amount
df.groupby(['Location','Category'])['Purchase Amount (USD)'].sum()
```

```python
#12. customer segment and their revenue
df.groupby('Subscription Status')['Purchase Amount (USD)'].sum()
```

```python
#13. categories with high purchase volume and low average purchase amount
categories = df.groupby('Category').agg({'Purchase Amount (USD)':['count','mean']}).reset_index()
categories
```

```python
categories.sort_values([('Purchase Amount (USD)','count'),('Purchase Amount (USD)','mean')],ascending=[False,True])
```

```python
#14. item purchased with strong average rating and low purchased volume
df2 = df.groupby('Item Purchased').agg({'Review Rating':'mean','Purchase Amount (USD)':'count'}).reset_index()
```

```python
df2.rename(columns={'Purchase Amount (USD)':'Purchase Volume'},inplace=True)
```

```python
df2.sort_values(by=['Review Rating','Purchase Volume'],ascending=[False,True])
```
